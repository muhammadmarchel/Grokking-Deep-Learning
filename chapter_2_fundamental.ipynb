## Eksplorasi Teoritis Mendalam (Theoretical Deep-Dive)

Bab kedua dari *Grokking Deep Learning* membawa kita keluar dari konsep abstrak dan masuk ke dalam inti matematika dari jaringan saraf tiruan (neural networks). Bab ini membongkar misteri "kecerdasan buatan" dengan menunjukkan bahwa pada tingkat yang paling dasar, sebuah jaringan saraf hanyalah sekumpulan operasi aritmetika—khususnya perkalian dan penjumlahan—yang diatur sedemikian rupa.

Tujuan utama dari bab ini adalah memahami anatomi dari sebuah prediksi. Sebelum kita bisa membuat mesin "belajar" (memperbaiki kesalahan), kita harus terlebih dahulu membuat mesin "berpikir" atau "memprediksi". Proses mesin menghasilkan prediksi dari data masukan disebut sebagai **Forward Propagation** (Propagasi Maju).

### 1. Anatomi Sebuah Prediksi: Input, Weight, dan Output
Untuk memahami *forward propagation*, kita harus memahami tiga komponen utama yang terlibat dalam proses ini:

*   **Input (Masukan):** Ini adalah data dunia nyata yang kita berikan kepada model. Input adalah hal yang kita ketahui dengan pasti. Dalam contoh di buku, input bisa berupa jumlah jari kaki (toes) seorang pemain bisbol, rata-rata kemenangan tim, atau jumlah penggemar. Komputer tidak memahami "pemain bisbol"; ia hanya memahami angka. Oleh karena itu, langkah pertama dalam deep learning selalu menerjemahkan objek dunia nyata menjadi angka.
*   **Weight (Bobot):** Ini adalah komponen yang paling krusial. Jika *input* adalah informasi yang kita miliki, maka *weight* adalah "pengetahuan" atau "ingatan" yang dimiliki oleh mesin. Bobot bertindak sebagai pengukur sensitivitas (volume knob).
    *   Jika sebuah bobot bernilai tinggi, artinya mesin menganggap input yang terhubung dengannya sangat penting untuk membuat prediksi.
    *   Jika bobot bernilai nol, mesin secara efektif mengabaikan input tersebut.
    *   Ketika kita mengatakan bahwa jaringan saraf "sedang belajar," yang sebenarnya terjadi secara harfiah adalah jaringan tersebut sedang **mengubah nilai bobot-bobotnya** secara matematis agar prediksi selanjutnya menjadi lebih akurat.
*   **Prediction / Output (Prediksi/Keluaran):** Ini adalah hasil akhir dari operasi matematika antara input dan weight. Ini adalah tebakan terbaik mesin berdasarkan informasi (input) dan pengetahuannya (weight) saat ini.

### 2. Mekanisme Forward Propagation: Sebuah Perkalian Sederhana
Dalam bentuknya yang paling murni dan paling sederhana (jaringan saraf dengan satu titik masukan dan satu titik keluaran), proses *forward propagation* dapat dirangkum dalam satu persamaan matematika yang elegan:

$$\text{Prediction} = \text{Input} \times \text{Weight}$$

Mengapa menggunakan perkalian? Karena perkalian bekerja secara proporsional.
*   Mengarahkan arah: Jika inputnya besar, prediksinya akan merespons menjadi besar (asumsi bobot positif).
*   Mengubah skala: Bobot ($w$) bertindak sebagai faktor skala (scaling factor). Jika kita ingin input mempengaruhi prediksi sebesar $20\%$, maka kita atur bobotnya menjadi $0.2$. Jika kita ingin input menurunkan nilai prediksi, kita gunakan bobot negatif (misalnya $-0.5$).

### 3. Otak Statis vs. Ingatan Dinamis (Static Code vs. Dynamic Weights)
Ada satu poin konseptual penting yang ditekankan dalam bab ini mengenai cara pandang kita terhadap kode program:
Dalam pemrograman tradisional, jika kita ingin membuat program lebih pintar, kita menambahkan lebih banyak baris kode (menambahkan `if`, `else`, `for loop`).

Namun, dalam *Deep Learning*, fungsi inti yang melakukan prediksi (yaitu `prediction = input * weight`) **hampir tidak pernah berubah**. Kode tersebut bersifat statis.
Lalu, bagaimana jaringannya bisa menjadi lebih cerdas seiring berjalannya waktu? Jawabannya ada pada nilai variabel `weight`. Kecerdasan tidak dikodekan dalam struktur programnya, melainkan disimpan di dalam *nilai angka pada bobot*. Ini adalah pergeseran pemikiran yang radikal: kita membangun struktur statis (jaringan), dan membiarkan data mengalir melewatinya untuk membentuk ingatan dinamis (bobot).

```python
# =====================================================================
# REPRODUKSI KODE BAB 2: MEMBUAT PREDIKSI SEDERHANA (FORWARD PROPAGATION)
# =====================================================================

# 1. Mendefinisikan Jaringan Saraf (The Neural Network Architecture)
# Fungsi ini mewakili seluruh arsitektur jaringan kita.
# Ia menerima dua parameter: 'input' (data yang kita ketahui) dan 'weight' (pengetahuan mesin).
def neural_network(input, weight):
    # Proses Forward Propagation:
    # Mesin mengambil data input, mengalikannya dengan bobot,
    # untuk mengekstrak seberapa penting data tersebut terhadap hasil akhir.
    prediction = input * weight
    return prediction

# 2. Inisialisasi Pengetahuan Mesin (Initializing the Weight)
# Ini adalah 'ingatan' dari model kita.
# Nilai 0.1 di sini biasanya pada awalnya dipilih secara acak (random initialization).
# Di bab-bab selanjutnya, nilai inilah yang akan diubah-ubah oleh mesin saat proses "belajar".
weight = 0.1

# 3. Menyiapkan Data Masukan (Input Data)
# Bayangkan ini adalah dataset historis.
# Daftar ini berisi rata-rata jumlah pertandingan (dalam satuan angka acak)
# dari berbagai tim bisbol yang telah kita kumpulkan.
number_of_toes = [8.5, 9.5, 10, 9]

# Kita mengambil observasi pertama (indeks ke-0) dari dataset kita untuk diuji.
input_data = number_of_toes[0]

# 4. Mengeksekusi Prediksi (Making the Prediction)
# Kita memasukkan data observasi dan bobot saat ini ke dalam jaringan saraf.
pred = neural_network(input_data, weight)

# 5. Menampilkan Hasil
print(f"Data Input (Observasi)  : {input_data}")
print(f"Bobot Jaringan (Weight) : {weight}")
print(f"Hasil Prediksi (Output) : {pred}")
```

### Penjelasan Eksekusi Kode:
1.  **`def neural_network(input, weight):`**: Baris ini adalah *cetak biru* (blueprint) dari jaringan kita. Meskipun terlihat seperti fungsi perkalian biasa, secara konseptual, ini merepresentasikan sebuah neuron buatan (artificial neuron). Neuron ini hanya tahu satu cara untuk merespons dunia: dengan mengambil sinyal masuk (`input`), menyesuaikan volumenya (`weight`), dan meneruskannya (`prediction`).
2.  **`weight = 0.1`**: Ini merepresentasikan status pengetahuan mesin pada *saat itu juga*. Sebelum mesin dilatih, ia tidak tahu apa-apa tentang bisbol atau prediksi. Jadi, kita harus memberinya titik awal. Pada praktiknya di dunia nyata, bobot awal ini diatur ke nilai acak (random).
3.  **`number_of_toes = [8.5, 9.5, ...]`**: Ini adalah simulasi dari sebuah *Dataset*. Dalam machine learning, kita jarang melakukan prediksi hanya pada satu angka. Kita memproses *batch* data. Di sini, kita mensimulasikan mengambil titik data pertama (`number_of_toes[0]`, yang bernilai `8.5`) untuk dilewatkan ke dalam jaringan.
4.  **`pred = neural_network(input_data, weight)`**: Ini adalah peristiwa eksekusinya. Saat fungsi ini dipanggil, matematika $\text{Prediction} = 8.5 \times 0.1$ dijalankan, menghasilkan nilai `0.85`. Mesin baru saja "berpikir" dan memberikan jawaban tebakannya.
