# Bab 7: Membayangkan Jaringan Saraf: Di Kepala Anda dan di Atas Kertas
*(How to Picture Neural Networks: In Your Head and on Paper)*

## Eksplorasi Teoritis Mendalam (Theoretical Deep-Dive)

Bab 7 berfungsi sebagai jembatan konseptual yang sangat penting. Setelah bergelut dengan matriks, *dot product*, dan algoritma *backpropagation* di bab-bab sebelumnya, bab ini menarik kita mundur sejenak untuk membangun **model mental visual** dari arsitektur jaringan saraf. Mengubah kode matematika menjadi gambar visual (nodes dan garis) membantu kita memahami intuisi di balik bagaimana informasi mengalir dan bagaimana korelasi ditangkap oleh mesin tanpa harus selalu membaca matriks angka.

### 1. Visualisasi Arsitektur: Nodes (Simpul) dan Edges (Garis)
Alih-alih hanya melihat *array* NumPy, jaringan saraf paling baik dipahami sebagai grafik berarah (*directed graph*).
*   **Nodes (Lingkaran):** Merepresentasikan titik data atau nilai aktivasi. Ini termasuk masukan (input), lapisan tersembunyi (hidden layers), dan tebakan akhir (output). Setiap *node* memegang satu nilai numerik.
*   **Edges (Garis):** Merepresentasikan bobot (*weights*). Secara visual, ketebalan, warna, atau ukuran garis ini dapat menunjukkan seberapa besar pengaruh (positif atau negatif) dari sebuah bobot.
*   **Mekanisme Visual:** *Forward propagation* dapat dibayangkan sebagai "air" (nilai input) yang mengalir melalui pipa (garis bobot). Pipa yang lebar (bobot besar) membiarkan banyak air lewat, sedangkan pipa yang tersumbat (bobot nol) menghentikan aliran air ke *node* berikutnya.

### 2. Intuisi di Balik Korelasi
Memvisualisasikan jaringan membantu kita memahami secara intuitif bagaimana mesin mendeteksi korelasi (hubungan sebab-akibat) dalam data.
*   Jika sebuah *input node* (misalnya: warna langit gelap) sering kali menyala bersamaan dengan *output* yang diharapkan (hujan), maka selama proses *Gradient Descent*, garis (bobot) yang menghubungkan keduanya akan tumbuh semakin tebal.
*   Sebaliknya, jika suatu input (misalnya: jumlah mobil merah yang lewat) tidak memiliki hubungan dengan hujan, garis bobotnya akan menyusut mendekati nol hingga menghilang. Ini adalah cara mesin membuang informasi yang tidak berguna secara visual.

### 3. Menerjemahkan Data Dunia Nyata ke dalam Matriks
Bab ini juga menekankan cara menyusun observasi dunia nyata menjadi struktur matematika yang bisa diproses mesin.
*   **Representasi Data:** Berdasarkan konvensi, setiap **baris (row)** dalam matriks data mewakili satu contoh atau observasi tunggal (misalnya, satu set status lampu lalu lintas di satu waktu). Setiap **kolom (column)** mewakili fitur spesifik (misalnya, lampu merah, kuning, atau hijau).
*   Memvisualisasikan *dataset* berdampingan dengan grafik jaringan membantu memperjelas bagaimana seluruh *batch* data (banyak baris sekaligus) dapat melewati jaringan secara bersamaan menggunakan operasi perkalian matriks yang sudah kita pelajari di Bab 3.

### 4. Memahami Overfitting dari Perspektif Visual
Ketika kita memvisualisasikan bobot jaringan, *overfitting* (kondisi di mana mesin terlalu menghafal data latih dan gagal di dunia nyata) dapat dilihat secara jelas. *Overfitting* divisualisasikan sebagai jaringan yang menciptakan jalur garis (bobot) yang sangat spesifik, kaku, dan rumit yang hanya berlaku sempurna untuk *dataset* saat ini, tetapi akan berantakan jika diberi input baru. Pemahaman visual ini menjadi fondasi krusial untuk mempelajari teknik *Regularization* (seperti *Dropout*) di bab selanjutnya.
