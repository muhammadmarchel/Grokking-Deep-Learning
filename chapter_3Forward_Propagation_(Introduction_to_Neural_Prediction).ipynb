## Eksplorasi Teoritis Mendalam (Theoretical Deep-Dive)

Pada bab sebelumnya, kita membangun model mental tentang jaringan saraf yang memproses satu titik data (misalnya: jumlah jari kaki) untuk menghasilkan satu prediksi (misalnya: peluang menang). Namun, di dunia nyata, fenomena kompleks jarang sekali hanya dipengaruhi oleh satu faktor tunggal. Bab ini memperluas konsep tersebut dengan memperkenalkan **vektor** dan **matriks** untuk memproses banyak variabel sekaligus.

### 1. Memprediksi dengan Banyak Masukan (Multiple Inputs)
Bayangkan memprediksi kemenangan sebuah tim bisbol. Anda tidak hanya melihat jumlah jari kaki pemain, tetapi juga rasio kemenangan sebelumnya (*win/loss record*) dan jumlah penggemar (*number of fans*).
*   **Weighted Sum (Jumlah Berbobot):** Saat jaringan menerima banyak masukan (multiple inputs), mesin akan mengalikan setiap input dengan bobotnya masing-masing, lalu menjumlahkan semua hasilnya. Ini disebut *weighted sum* atau *dot product* (produk titik).
*   **Makna Intuisi:** *Dot product* pada dasarnya adalah pengukur kemiripan (similarity). Operasi ini menggabungkan berbagai "potongan" informasi yang berbeda menjadi satu kesimpulan utuh yang lebih akurat daripada jika hanya mengandalkan satu masukan saja.

### 2. Memprediksi dengan Banyak Keluaran (Multiple Outputs)
Sebaliknya, satu informasi (input) bisa digunakan untuk memprediksi banyak hal yang berbeda. Misalnya, dari persentase kemenangan tim, kita bisa memprediksi: 1) peluang pemain cedera, 2) peluang menang lagi, dan 3) tingkat kesedihan tim.
*   **Mekanisme:** Jaringan mengambil satu *input* dan mengalikannya dengan *banyak bobot (weights)* yang berbeda. Setiap bobot difokuskan untuk menghasilkan satu tebakan spesifik.

### 3. Memprediksi dengan Banyak Masukan dan Keluaran (Multiple Inputs & Outputs)
Ketika kita menggabungkan konsep 1 dan 2, kita mendapatkan jaringan saraf sejati yang penuh dengan koneksi silang. Setiap input terhubung ke setiap output. Operasi untuk menyelesaikan ini tidak lagi menggunakan perkalian angka biasa, melainkan **Perkalian Matriks (Matrix Multiplication)**. Matriks hanyalah kumpulan angka yang disusun dalam baris dan kolom.

### 4. Arsitektur Jaringan Mendalam (Predicting on Predictions)
Di sinilah kata "Deep" dalam Deep Learning berasal. Kita tidak langsung memetakan input menjadi output akhir.
*   **Hidden Layers (Lapisan Tersembunyi):** Input dikalikan dengan matriks bobot pertama untuk menghasilkan prediksi perantara (disebut *hidden layer*).
*   **Hierarki Pembelajaran:** Prediksi perantara ini kemudian digunakan sebagai *input baru* untuk matriks bobot kedua demi menghasilkan prediksi akhir. Ini memungkinkan jaringan untuk memahami data dalam tingkatan hierarki (misalnya: piksel -> garis -> bentuk hidung -> wajah manusia).

```python
import numpy as np

# =====================================================================
# REPRODUKSI KODE BAB 3: PREDIKSI DENGAN MULTIPLE INPUTS & OUTPUTS
# Menggunakan NumPy untuk efisiensi matematika matriks (Vectorization)
# =====================================================================

# ---------------------------------------------------------
# SKENARIO 1: Multiple Inputs -> Single Output (Dot Product)
# ---------------------------------------------------------
# Bobot masing-masing input: (toes, wlrec, nfans)
weights_single = np.array([0.1, 0.2, 0.0])

def neural_network_single(input_vector, weight_vector):
    # Fungsi .dot() dari NumPy melakukan operasi 'Weighted Sum' secara otomatis
    prediction = input_vector.dot(weight_vector)
    return prediction

# Dataset historis kita (3 variabel berbeda)
toes = np.array([8.5, 9.5, 9.9, 9.0])   # Jumlah jari kaki
wlrec = np.array([0.65, 0.8, 0.8, 0.9]) # Rasio menang/kalah
nfans = np.array([1.2, 1.3, 0.5, 1.0])  # Jumlah fans dalam jutaan

# Mengambil data di baris/indeks ke-0 sebagai input
input_data = np.array([toes[0], wlrec[0], nfans[0]])

pred_single = neural_network_single(input_data, weights_single)
print(f"Input: {input_data}")
print(f"Skenario 1 - Hasil Prediksi (Peluang Menang): {pred_single:.4f}\n")


# ---------------------------------------------------------
# SKENARIO 2: Deep Learning Dasar (Predicting on Predictions)
# ---------------------------------------------------------
# Matriks Bobot Layer 1 (Input ke Hidden Layer)
# Baris mewakili Node (Output Perantara), Kolom mewakili Input
ih_wgt = np.array([
    [0.1, 0.2, -0.1], # Bobot pembentuk hidden node ke-1
    [-0.1, 0.1, 0.9], # Bobot pembentuk hidden node ke-2
    [0.1, 0.4, 0.1]   # Bobot pembentuk hidden node ke-3
])

# Matriks Bobot Layer 2 (Hidden Layer ke Final Output)
hp_wgt = np.array([
    [0.3, 1.1, -0.3], # Bobot pembentuk prediksi akhir: Cedera (Hurt)
    [0.1, 0.2, 0.0],  # Bobot pembentuk prediksi akhir: Menang (Win)
    [0.0, 1.3, 0.1]   # Bobot pembentuk prediksi akhir: Sedih (Sad)
])

# Menyimpan semua matriks bobot di dalam list
weights_deep = [ih_wgt, hp_wgt]

def neural_network_deep(input_vector, weights):
    # Langkah 1: Input -> Hidden Layer
    hidden_layer_pred = input_vector.dot(weights[0].T) # Gunakan Transpose (.T) agar dimensi sesuai

    # Langkah 2: Hidden Layer -> Final Output
    final_pred = hidden_layer_pred.dot(weights[1].T)
    return final_pred

pred_deep = neural_network_deep(input_data, weights_deep)
print(f"Skenario 2 - Hasil Prediksi Deep Network:")
print(f"Peluang Cedera : {pred_deep[0]:.4f}")
print(f"Peluang Menang : {pred_deep[1]:.4f}")
print(f"Peluang Sedih  : {pred_deep[2]:.4f}")
```

### Penjelasan Eksekusi Kode:

1.  **Mengapa `import numpy as np`?** NumPy adalah fondasi komputasi data sains di Python. Melakukan *looping* menggunakan `for` pada Python biasa (seperti `for i in range(len(inputs)): result += inputs[i] * weights[i]`) sangatlah lambat untuk data berskala besar. NumPy menjalankan operasi ini di bawah kap menggunakan bahasa C yang teroptimasi, membuatnya jauh lebih efisien (*Vectorization*).
2.  **`input_vector.dot(weight_vector)`:** Ini adalah "Sihir" dari *Dot Product* (Produk Titik). Jaringan ini secara matematis menghitung: `(8.5 * 0.1) + (0.65 * 0.2) + (1.2 * 0.0)`. Hasil dari penjumlahan bobot ini disatukan menjadi satu prediksi akhir `0.98`.
3.  **Matriks Bobot (`ih_wgt` & `hp_wgt`):** Dalam arsitektur yang lebih dalam (*Deep Learning*), kita menyusun array di dalam array (Matriks). Tiap baris dalam matriks `ih_wgt` bertugas mencari "pola tertentu" dari ketiga data input tersebut.
4.  **`weights[0].T` (Transpose):** Saat kita memutar arsitektur matriks menggunakan `.T` (*Transpose*), kita menyejajarkan baris dan kolom agar sesuai dengan operasi aljabar linear standar.
5.  **`hidden_layer_pred = ...` lalu `final_pred = ...`:** Inilah struktur "Deep" yang sebenarnya. Prediksi tidak dilakukan dalam satu tembakan. Data input diubah menjadi "representasi tersembunyi" (hidden layer) terlebih dahulu. Kemudian, tebakan tersembunyi ini digunakan sebagai input nyata untuk membuat keputusan final (Cedera, Menang, atau Sedih).
