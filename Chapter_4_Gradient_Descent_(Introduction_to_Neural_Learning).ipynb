## Eksplorasi Teoritis Mendalam (Theoretical Deep-Dive)

Pada bab-bab sebelumnya, jaringan saraf kita hanya bertindak sebagai kalkulator statis. Bab ini akhirnya memperkenalkan mekanisme "Belajar" (Learning). Belajar bagi sebuah mesin pada dasarnya adalah proses mengoptimalkan bobot (*weights*) agar kesalahan (*error*) dari prediksinya sekecil mungkin. Teknik utama yang digunakan untuk melakukan ini disebut **Gradient Descent**.

### 1. Siklus Inti Machine Learning: Predict, Compare, Learn
Proses mesin belajar selalu mengikuti tiga langkah abadi ini:
*   **Predict (Prediksi):** Jaringan mengambil input dan mengalikannya dengan bobot untuk menghasilkan tebakan.
*   **Compare (Bandingkan):** Jaringan membandingkan hasil tebakannya dengan jawaban yang sebenarnya (target/label). Selisih antara keduanya disebut *Error*.
*   **Learn (Belajar):** Jaringan menggunakan informasi *Error* tersebut untuk mengoreksi atau mengubah nilai bobotnya agar tebakan di masa depan lebih akurat.

### 2. Mengukur Kesalahan (Mean Squared Error)
Mengapa kita tidak hanya menggunakan selisih langsung seperti `(prediksi - target)` untuk mengukur error? Mengapa harus dikuadratkan menjadi `(prediksi - target) ** 2`?
*   **Arah vs Kuantitas:** Error murni bisa bernilai negatif atau positif. Jika kita mencoba meminimalkan angka negatif yang tidak terbatas, algoritma akan rusak. Mengkuadratkannya memastikan error selalu positif (absolut).
*   **Penalti Berlebihan:** Pengkuadratan memberikan "hukuman" (penalti) yang lebih besar untuk error yang sangat meleset, dan penalti yang sangat kecil untuk tebakan yang hampir benar. Ini membentuk kurva mangkuk (bentuk-U) yang sempurna, yang membuat pencarian nilai terendah (titik dasar mangkuk) menjadi lebih mudah dihitung secara matematis.

### 3. Gradient Descent vs. Hot and Cold Learning
*   *Hot and Cold Learning:* Mencoba menebak dengan menggeser bobot sedikit ke atas dan ke bawah, lalu melihat mana yang menghasilkan error lebih kecil. Ini sangat tidak efisien dan lambat.
*   *Gradient Descent:* Menghitung *turunan/gradien (derivative)* secara presisi menggunakan kalkulus. Kita menghitung `arah` dan `jumlah` perubahan bobot dalam satu langkah matematika tunggal. Dengan mengalikan error dengan input `(prediksi - target) * input`, kita mendapatkan angka pasti seberapa banyak bobot harus digeser.

### 4. Parameter 'Alpha' (Learning Rate)
Saat kita menghitung seberapa banyak bobot harus digeser, seringkali pergeserannya terlalu besar, menyebabkan jaringan melompat melewati titik optimal dan error malah meledak (*overshooting*). Di sinilah `alpha` (atau *learning rate*) masuk. Ini adalah angka kecil (misalnya `0.01` atau `0.1`) yang digunakan untuk memperlambat langkah penyesuaian bobot.

```python
# =====================================================================
# REPRODUKSI KODE BAB 4: GRADIENT DESCENT PADA SATU BOBOT
# Jaringan Saraf yang belajar memperbaiki kesalahannya
# =====================================================================

# 1. Inisialisasi Data & Bobot
weight = 0.5            # Pengetahuan awal (tebakan sembarangan)
goal_pred = 0.8         # Target prediksi yang INGIN kita capai (Kebenaran)
input_data = 0.5        # Data observasi
alpha = 0.1             # Learning Rate (mencegah langkah update yang terlalu besar)

# 2. Proses Belajar (Looping Iterasi Gradient Descent)
print(f"Target yang ingin dicapai: {goal_pred}\n")
print(f"{'Iterasi':<10} | {'Prediksi':<12} | {'Error':<12} | {'Bobot Baru'}")
print("-" * 55)

for iteration in range(20): # Kita paksa mesin belajar selama 20 kali percobaan

    # LANGKAH 1: PREDICT (Prediksi)
    pred = input_data * weight

    # LANGKAH 2: COMPARE (Bandingkan / Hitung Error)
    # Mean Squared Error
    error = (pred - goal_pred) ** 2

    # LANGKAH 3: LEARN (Belajar / Kalkulasi Arah & Besaran Koreksi)
    # Ini disebut 'Delta' atau 'Direction and Amount'.
    # Menghitung selisih murni antara tebakan dan jawaban
    delta = pred - goal_pred

    # Menghitung Turunan (Derivative/Gradient): Input * Delta
    weight_delta = input_data * delta

    # UPDATE BOBOT KITA: Kurangi bobot lama dengan (gradient * learning rate)
    weight = weight - (alpha * weight_delta)

    # Menampilkan progress di setiap iterasi
    print(f"Iterasi {iteration:<2} | Prediksi: {pred:.4f} | Error: {error:.4f} | W = {weight:.4f}")

print("\nSukses! Mesin berhasil mengubah bobotnya sendiri sehingga tebakannya cocok dengan target.")
```

### Penjelasan Eksekusi Kode:

1.  **Looping Pembelajaran (`for iteration in range(20):`):** Proses mesin belajar tidak terjadi dalam satu kali tembakan. Kita harus melakukan siklus *Forward Propagation* (memprediksi) dan *Backpropagation/Gradient Descent* (mengoreksi) berkali-kali hingga error menyentuh angka `0`.
2.  **`delta = pred - goal_pred`:** Ini adalah fondasi dari *Gradient Descent*. Operasi ini mengembalikan nilai absolut dari kesalahan. Jika `pred` terlalu tinggi, `delta` akan positif (menyuruh bobot untuk turun). Jika `pred` terlalu rendah, `delta` akan negatif (menyuruh bobot untuk naik).
3.  **`weight_delta = input_data * delta`:** Ini adalah rumus gradien/turunan. Mengapa kita harus mengalikan `delta` dengan `input_data`?
    *   **Scaling:** Jika inputnya besar, perubahan bobotnya juga harus besar agar cepat sampai ke tujuan.
    *   **Negative Reversal:** Jika inputnya negatif, arah `delta` harus dibalik agar bobot bergerak ke arah yang benar.
    *   **Stopping:** Jika `input` adalah nol (0), maka tebakan mesin tidak memengaruhi output sama sekali, sehingga tidak ada yang perlu dipelajari (bobot tidak akan diubah).
4.  **`weight = weight - (alpha * weight_delta)`:** Ini adalah momen di mana pembaruan bobot terjadi. Penggunaan minus (`-`) berfungsi untuk bergerak berlawanan dengan arah gradient (bergerak menuju lembah atau titik error terendah). Faktor `alpha` memastikan mesin mengambil langkah-langkah kecil untuk menghindari *overshooting* (melebihi target dan terpantul bolak-balik).
